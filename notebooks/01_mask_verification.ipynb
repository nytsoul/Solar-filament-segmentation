# Mask Verification
This notebook demonstrates loading an image and its annotations, and visualizing the masks.

```python
import os
import sys
import cv2
import matplotlib.pyplot as plt
import numpy as np

sys.path.append(os.path.dirname(os.path.abspath('')))

from src.data.annotations import MAGFiLOAnnotationParser
from src.data.masks import create_individual_mask, create_instance_masks, create_semantic_mask
```

```python
# Load parser
json_path = '../MAGFiLO_1.0_Kaggle_2026/train/MAGFiLO_1.0_Annotations_kaggle2026_train.json'
parser = MAGFiLOAnnotationParser(json_path)
print("Loaded parser")
```

```python
# Select a real image
img_id = list(parser.img_id_to_filename.keys())[0]
filename = parser.img_id_to_filename[img_id]
img_path = os.path.join('../MAGFiLO_1.0_Kaggle_2026/train/train_images', filename)

img = cv2.imread(img_path, cv2.IMREAD_COLOR)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
annotations = parser.get_annotations_for_image(img_id)

print(f"Image: {filename}")
print(f"Filament count: {len(annotations)}")
```

```python
# Reconstruct masks
h, w = img.shape[:2]
instance_mask = create_instance_masks(annotations, h, w)
semantic_mask = create_semantic_mask(annotations, h, w)

total_pixels = np.count_nonzero(semantic_mask)
print(f"Total filament pixels: {total_pixels}")
```

```python
# Display original image
plt.figure(figsize=(10, 10))
plt.imshow(img_rgb)
plt.title('Original H-alpha Image')
plt.axis('off')
plt.show()
```

```python
# Display individual instances
plt.figure(figsize=(10, 10))
plt.imshow(instance_mask, cmap='jet')
plt.title('Individual Instances')
plt.axis('off')
plt.show()
```

```python
# Display combined semantic mask
plt.figure(figsize=(10, 10))
plt.imshow(semantic_mask, cmap='gray')
plt.title('Combined Semantic Mask')
plt.axis('off')
plt.show()
```

```python
# Display overlay
overlay = img_rgb.copy()
overlay[semantic_mask == 1] = [255, 0, 0] # Red overlay
plt.figure(figsize=(10, 10))
plt.imshow(overlay)
plt.title('Overlay')
plt.axis('off')
plt.show()
```
